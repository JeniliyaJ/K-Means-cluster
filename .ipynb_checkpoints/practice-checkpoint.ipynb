```python
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
heights=np.array([150,152,154,160,166,200,220,300,350,400,420,500,600,700,750,650])
```

```python
heights=heights.reshape(-1,1)
```

 ## apply k means cluster algorithmn

```python
kmeans=KMeans(n_clusters=3,random_state=0)
```

```python
kmeans.fit(heights)
```

```python
label=kmeans.labels_
```

```python
kmeans.cluster_centers_
```

## plot the cluster

```python
plt.scatter(heights, np.zeros_like(heights),s=100)
plt.scatter(kmeans.cluster_centers_,[0,0,0],marker='X',s=100)
plt.grid(True)
```

```python
for i,h in enumerate(heights):
    print(f"height{h[0]} lable is {label[i]}")
```
